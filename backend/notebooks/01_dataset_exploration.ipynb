# Phase 1: Customer Support on Twitter Exploration & @AppleSupport Selection

This notebook explores the Kaggle Twitter Customer Support (`thoughtvector/customer-support-on-twitter`) dataset and performs rigorous profiling on the selected brand `@AppleSupport`.

```python
import os
import pandas as pd
import numpy as np

data_path = '../data/samples/apple_support_clean.parquet'
df = pd.read_parquet(data_path)
print(f"Loaded clean conversation pairs: {len(df):,}")
df.head(3)
```

## Message Length & Dialogue Dynamics
Examine customer message length versus brand diagnostic length.

```python
df['customer_words'] = df['customer_text'].astype(str).apply(lambda x: len(x.split()))
df['brand_words'] = df['brand_reply_text'].astype(str).apply(lambda x: len(x.split()))

print("Customer message word statistics:")
print(df['customer_words'].describe())
print("\nBrand reply word statistics:")
print(df['brand_words'].describe())
```

## Actionable Documentation Link Rate
Check how often @AppleSupport grounds replies in official `apple.com` documentation.

```python
apple_links = df['brand_reply_text'].str.contains('apple.com|t.co', case=False).mean()
print(f"Proportion of replies with links / KB pointers: {apple_links * 100:.1f}%")
```
